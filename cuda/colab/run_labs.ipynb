```python
%%bash
cd /content/cuda-labs
mkdir -p bin
for f in kernels/*.cu; do
  n=$(basename $f .cu)
  echo building $n
  nvcc -O3 --use_fast_math -std=c++14 -arch=native $f -o bin/$n || exit 1
done
echo BUILD_OK
```

```python
import subprocess, os
labs = sorted(l for l in os.listdir('/content/cuda-labs/bin') if l.startswith('lab'))
fail = []
for lab in labs:
    print('='*72); print(lab)
    r = subprocess.run(['/content/cuda-labs/bin/'+lab], capture_output=True, text=True)
    print(r.stdout[-3000:])
    if r.returncode != 0:
        fail.append(lab); print('STDERR:', r.stderr[-800:])
print('
FAILED:', fail if fail else 'NONE - ALL LABS PASS')
```
